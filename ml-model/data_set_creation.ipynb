```python
#calculate g-force
import numpy as np

combined_df['g_force'] = np.sqrt(
    combined_df['Accelerometer(x)']**2 +
    combined_df['Accelerometer(y)']**2 +
    combined_df['Accelerometer(z)']**2
) / 9.8
```

```python
combined_df
```

```python
combined_df['acceleration'] = np.sqrt(
    combined_df['Accelerometer(x)']**2 +
    combined_df['Accelerometer(y)']**2 +
    combined_df['Accelerometer(z)']**2
)   
```

```python
# Calculate the change in acceleration
combined_df['accel_change'] = combined_df['acceleration'].diff()
```

```python
combined_df
```

```python
combined_df['accel_magnitude'] = np.sqrt(
    combined_df['Accelerometer(x)']**2 +
    combined_df['Accelerometer(y)']**2 +
    combined_df['Accelerometer(z)']**2
)

# Gyroscope Magnitude
combined_df['gyro_magnitude'] = np.sqrt(
    combined_df['gyroscope(x)']**2 +
    combined_df['gyroscope(y)']**2 +
    combined_df['gyroscope(z)']**2
)
```

```python
combined_df.columns
```

```python
# resample data into 1 second windows
combined_df['time'] = pd.to_datetime(combined_df['time'])
combined_df
```

```python
combined_df.shape
```

```python
# First set time as index
combined_df.set_index('time', inplace=True)

# Define your aggregation rules
agg_rules = {
    'accel_magnitude': 'std',
    'gyro_magnitude': 'std',
    # For all other columns (excluding 'time'), apply max
    'Accelerometer(seconds_elapsed)': 'max',
    'Accelerometer(z)': 'max',
    'Accelerometer(y)': 'max',
    'Accelerometer(x)': 'max',
    'gyroscope(seconds_elapsed)': 'max',
    'gyroscope(z)': 'max',
    'gyroscope(y)': 'max',
    'gyroscope(x)': 'max',
    'g_force': 'max',
    'acceleration': 'max',
    'accel_change': 'max'
}

# Perform the resampling and aggregation
windowed_df = combined_df.resample('1S').agg(agg_rules).dropna(how='all').reset_index()

```

```python

def classify_road(accel_std, gyro_std, accel_threshold=0.5, gyro_threshold=0.5):
    if accel_std < accel_threshold and gyro_std < gyro_threshold:
        return 'Normal'
    else:
        return 'Rough'

# Apply classification
windowed_df['road_type'] = windowed_df.apply(
    lambda row: classify_road(row['accel_magnitude'], row['gyro_magnitude']), axis=1
)
```

```python
import pandas as pd
from langchain.prompts import PromptTemplate
from langchain_openai import AzureChatOpenAI 
from langchain.chains import LLMChain

# 2. Prompt Template with rules and few-shot examples
template = """
You are an expert in analyzing driving behavior based on vehicle sensor data.

Classify the driving event as either "Safe" or "Dangerous" based on the following 4 parameters:
- g_force: overall G-force acting on the vehicle
- accel_change: change in acceleration (m/s²)
- road_type: can be "Normal" or "Rough"
- gyro_z_abs: absolute gyroscope Z-axis rotation in rad/s (indicating turn sharpness)

Use the following rules:
- If g_force > 1.0 → likely Dangerous
- If accel_change > 1.4 → possible harsh acceleration or braking → Dangerous
- If gyro_z_abs > 0.5 → likely a sharp turn → Dangerous
- If road_type is Rough and any of the above conditions are met → more likely Dangerous

Examples:

Example 1:
g_force: 0.85
accel_change: 0.42
road_type: Normal
gyro_z_abs: 0.35
Label: Safe

Example 2:
g_force: 1.25
accel_change: 2.8
road_type: Rough
gyro_z_abs: 1.3
Label: Dangerous

Example 3:
g_force: 0.95
accel_change: 1.0
road_type: Rough
gyro_z_abs: 0.8
Label: Dangerous

Example 4:
g_force: 0.72
accel_change: 0.3
road_type: Normal
gyro_z_abs: 0.25
Label: Safe

Now classify this event:

g_force: {g_force}
accel_change: {accel_change}
road_type: {road_type}
gyro_z_abs: {gyro_z_abs}
Label:
"""

# 3. Create the PromptTemplate
prompt = PromptTemplate(
    input_variables=["g_force", "accel_change", "road_type", "gyro_z_abs"],
    template=template,
)
OPENAI_API_VERSION="2024-06-01"
AZURE_OPENAI_ENDPOINT="https://openai-charliedev.openai.azure.com/"
AZURE_OPENAI_API_KEY="3cf6a02fbd8d4adf8d7e03d84873e538"
AZURE_OPENAI_DEPLOYMENT="gpt-4o-mini"
AZURE_OPENAI_API_TYPE="azure"
# 4. Load the LLM
llm = AzureChatOpenAI(openai_api_version="2024-06-01",
                      azure_endpoint = AZURE_OPENAI_ENDPOINT,
                      openai_api_key=AZURE_OPENAI_API_KEY,
                      openai_api_type = AZURE_OPENAI_API_TYPE,
                      deployment_name = AZURE_OPENAI_DEPLOYMENT,)

import time

chain = LLMChain(llm=llm, prompt=prompt)

# Retry logic with a maximum of 10 retries and a delay of 2 seconds between retries
def run_chain_with_retry(inputs, max_retries=10, delay=2):
    attempt = 0
    while attempt < max_retries:
        try:
            # Attempt to run the chain
            result = chain.run(inputs)
            return result  # Return the result if successful
        except Exception as e:
            attempt += 1
            print(f"Attempt {attempt} failed: {e}")
            if attempt < max_retries:
                time.sleep(delay)  # Wait for the specified delay before retrying
            else:
                raise Exception("Maximum retry attempts reached.")  # Raise exception after max retries

# Iterate through each row and get the label
labels = []
for idx, row in final_df.iterrows():
    inputs = {
        "g_force": row["g_force"],
        "accel_change": row["accel_change"],
        "road_type": row["road_type"],
        "gyro_z_abs": row["gyro_z_abs"]
    }

    try:
        result = run_chain_with_retry(inputs)
        label = result.strip().split("Label:")[-1].strip()
    except Exception as e:
        label = "ERROR"
        print(f"Failed on row {idx}: {e}")

    labels.append(label)

# Add labels to DataFrame
final_df["label"] = labels

# Save or use the labeled DataFrame
final_df.to_csv("labeled_driving_data.csv", index=False)
print(final_df.head())

```

```python
import re

def clean_label(label):
    # Lowercase the label and remove non-alphabet characters
    label = re.sub(r'[^a-zA-Z]', '', label.lower())
    if 'safe' in label:
        return 'Safe'
    elif 'dangerous' in label:
        return 'Dangerous'


# Apply the function to the 'label' column
df['label'] = df['label'].apply(clean_label)

# Check result
print(df['label'].value_counts())
```
